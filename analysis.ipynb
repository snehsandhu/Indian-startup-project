```python
df.drop(columns = ['Remarks'], inplace = True)
```

```python
df.set_index('Sr No', inplace = True)
```

```python
df.rename(columns = {
    'Sr  No':'sr no',
    'Date dd/mm/yyyy': 'date',
    'Startup Name': 'startup',
    'Industry Vertical':'Vertical',
    'SubVertical':'subvertical',
    'City  Location':'city ',
    'Investors Name':'investors',
    'InvestmentnType': 'round',
    'Amount in USD': 'amount in usd'
}, inplace = True)
```

## changing amount to float


```python
df1 = df
```

```python
df1['amount in usd'] = df1['amount in usd'].str.replace(',','')
df1['amount in usd'] = df1['amount in usd'].str.replace('undisclosed','0')
df1['amount in usd'] = df1['amount in usd'].str.replace('unknown','0')
df1['amount in usd'] = df1['amount in usd'].str.replace('Undisclosed','0')
```

```python
df1['amount in usd'] = df1['amount in usd'].fillna(0)
```

```python
df1['amount in usd'] = pd.to_numeric(
    df1['amount in usd'],
    errors='coerce'
)
```

```python
df['amount in usd']= df1['amount in usd'].astype('float')
```

```python
df.info()
```

## Changing date to DateTime

```python
df1["date"] = pd.to_datetime(
    df1["date"],
    format="%d/%m/%Y",  errors='coerce'
)
```

```python
df["date"]= df1["date"]
```

```python
df.info()
```

## droping null values as much possible

```python
df = df.dropna(subset=['date','startup','Vertical','investors','city ','round','amount in usd'] )
```

```python
df.info()
```

```python
df.to_csv("startup_cleaned_data.csv", index = False)
```
